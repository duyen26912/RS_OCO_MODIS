## Gộp dữ liệu file OCO2 và OCO3

```python
import pandas as pd

# Đọc 2 file CSV
df1 = pd.read_csv(r"E:\DownloadData\co2_nasa\data_processed\csv_yearly\2024\oco2\oco2_vn_qual0_2024.csv")
df2 = pd.read_csv(r"E:\DownloadData\co2_nasa\data_processed\csv_yearly\2024\oco3\oco3_vn_qual0_2024.csv")

# Gộp 2 file
df_merged = pd.concat([df1, df2], ignore_index=True)

# Chuyển cột 'file_date' về dạng datetime (tự động nhận dạng định dạng ngày)
df_merged['date'] = pd.to_datetime(df_merged['date'], errors='coerce')

# Kiểm tra nếu có dòng lỗi định dạng ngày
invalid_dates = df_merged[df_merged['date'].isna()]
if not invalid_dates.empty:
    print("⚠️ Có dòng lỗi định dạng ngày, sẽ bị xoá:")
    print(invalid_dates)

# Xoá các dòng có giá trị NaT ở 'file_date'
df_merged = df_merged.dropna(subset=['date'])

# Sắp xếp theo ngày tăng dần
df_merged = df_merged.sort_values(by='date').reset_index(drop=True)

# Xuất file CSV đã gộp và sắp xếp
df_merged.to_csv(r"E:\DownloadData\co2_nasa\data_processed\merge_oco2_oco3\oco_merged_2024.csv", index=False)

print("✅ Đã gộp và sắp xếp xong dữ liệu!")
```

```python
import pandas as pd

# Đọc 2 file CSV
df1 = pd.read_csv(r"E:\DownloadData\co2_nasa\data_processed\csv_yearly\2024\oco2\oco2_vn_qual0_2024.csv")
df2 = pd.read_csv(r"E:\DownloadData\co2_nasa\data_processed\csv_yearly\2024\oco3\oco3_vn_qual0_2024.csv")

# Đổi cột date sang datetime
df1['date'] = pd.to_datetime(df1['date'], errors='coerce')
df2['date'] = pd.to_datetime(df2['date'], errors='coerce')

# Lấy các giá trị date hợp lệ (không NaT)
dates1 = set(df1['date'].dropna())
dates2 = set(df2['date'].dropna())

# Tìm ngày trùng nhau giữa hai file
common_dates = sorted(dates1 & dates2)

if len(common_dates) == 0:
    print("✅ Không có ngày nào trùng giữa hai file.")
else:
    print(f"⚠️ Có {len(common_dates)} ngày trùng giữa hai file:")
    for d in common_dates:
        print(d.date())  # in dạng YYYY-MM-DD

    # Nếu muốn xem chi tiết các dòng tương ứng:
    print("\n📌 Các dòng trong OCO2 có date trùng:")
    print(df1[df1['date'].isin(common_dates)].head())

    print("\n📌 Các dòng trong OCO3 có date trùng:")
    print(df2[df2['date'].isin(common_dates)].head())
```
