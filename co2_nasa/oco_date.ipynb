```python
import pandas as pd

# 🔹 Đường dẫn đến file CSV lớn
input_csv = r"E:\DownloadData\co2_nasa\data_processed\rice_method_3\2021\oco2_filtered_rice_2021.csv"

# 🔹 Đọc dữ liệu
df = pd.read_csv(input_csv)

# 🔹 Lấy các giá trị ngày duy nhất trong cột 'file_date'
unique_dates = df["file_date"].dropna().unique()

# 🔹 Sắp xếp lại cho gọn
unique_dates = sorted(unique_dates)

# 🔹 In ra màn hình
for d in unique_dates:
    print(d)

# 🔹 (Tuỳ chọn) Lưu danh sách ngày ra file CSV riêng
output_csv = r"E:\DownloadData\co2_nasa\data_processed\date_list\dates_2021.csv"
pd.DataFrame({"file_date": unique_dates}).to_csv(output_csv, index=False)
print(f"\n✅ Đã lưu danh sách ngày vào: {output_csv}")
```
