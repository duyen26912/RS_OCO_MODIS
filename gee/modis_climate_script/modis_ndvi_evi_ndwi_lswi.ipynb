# BƯỚC 4: ĐỊNH NGHĨA HÀM

```python
countries = ee.FeatureCollection("FAO/GAUL/2015/level0")
vietnam = countries.filter(ee.Filter.eq('ADM0_NAME', 'Viet Nam')).first()
vietnam_geometry = vietnam.geometry()
print("✅ Đã load ranh giới Việt Nam")
```

```python
def mask_mod09ga_quality(image):
    qc_500m = image.select('QC_500m')
    modland_qa = qc_500m.bitwiseAnd(0x0003)
    ideal_quality_mask = modland_qa.eq(0)
    return image.updateMask(ideal_quality_mask)
```

```python
def calculate_ndvi(img):
    return img.normalizedDifference(['nir', 'red']).rename('NDVI')

def calculate_evi(img):
    nir_evi = img.select('nir')
    red_evi = img.select('red')
    blue_evi = img.select('blue')
    evi_numerator = nir_evi.subtract(red_evi)
    evi_denominator = nir_evi.add(red_evi.multiply(6)).subtract(blue_evi.multiply(7.5)).add(1)
    return evi_numerator.divide(evi_denominator).multiply(2.5).rename('EVI')

def calculate_lswi(img):
    return img.normalizedDifference(['nir', 'swir1']).rename('LSWI')

def calculate_ndwi_mcfeeters(img):
    return img.normalizedDifference(['green', 'nir']).rename('NDWI_McFeeters')
```

```python
def export_to_drive(image, date_string, folder_name):
    output_name = f'MOD09GA_Vietnam_NDVI_EVI_LSWI_NDWI_{date_string}'

    task = ee.batch.Export.image.toDrive(
        image=image,
        description=output_name,
        folder=folder_name,
        fileNamePrefix=output_name,
        region=vietnam_geometry.bounds(),
        scale=500,
        crs='EPSG:4326',
        maxPixels=1e13,
        fileFormat='GeoTIFF'
    )

    task.start()
    print(f'✓ Đã tạo task: {output_name}')
    return task
```

```python
def process_date(date_string, folder_name):
    print(f'\n{"="*60}')
    print(f'Xử lý ngày: {date_string}')
    print(f'{"="*60}')

    start_date = f'{date_string}T00:00:00'
    end_date = f'{date_string}T23:59:59'

    mod09ga_collection = ee.ImageCollection('MODIS/061/MOD09GA')
    mod09ga = mod09ga_collection.filterDate(start_date, end_date).filterBounds(vietnam_geometry)

    image_count = mod09ga.size().getInfo()
    print(f'Số ảnh tìm thấy: {image_count}')

    if image_count == 0:
        print(f'⚠️  Không có dữ liệu cho {date_string}')
        return False

    print('Đang xử lý...')
    raw_mosaic = mod09ga.mosaic().clip(vietnam_geometry)
    image = mask_mod09ga_quality(raw_mosaic)

    blue_band = image.select('sur_refl_b03').multiply(0.0001).rename('blue')
    green_band = image.select('sur_refl_b04').multiply(0.0001).rename('green')
    red_band = image.select('sur_refl_b01').multiply(0.0001).rename('red')
    nir_band = image.select('sur_refl_b02').multiply(0.0001).rename('nir')
    swir1_band = image.select('sur_refl_b06').multiply(0.0001).rename('swir1')

    processed_image = ee.Image.cat([blue_band, green_band, red_band, nir_band, swir1_band])

    print('Đang tính các chỉ số...')
    ndvi_image = calculate_ndvi(processed_image)
    evi_image = calculate_evi(processed_image)
    lswi_image = calculate_lswi(processed_image)
    ndwi_mcfeeters_image = calculate_ndwi_mcfeeters(processed_image)

    final_image = ee.Image.cat([
        ndvi_image, evi_image, lswi_image, ndwi_mcfeeters_image
    ]).toFloat()

    try:
        export_to_drive(final_image, date_string, folder_name)
        return True
    except Exception as e:
        print(f'✗ Lỗi: {str(e)}')
        return False
```

```python
def check_tasks_status():
    tasks = ee.batch.Task.list()
    status_count = {
        'READY': 0, 'RUNNING': 0, 'COMPLETED': 0,
        'FAILED': 0, 'CANCELLED': 0
    }

    for task in tasks[:100]:
        state = task.status()['state']
        if state in status_count:
            status_count[state] += 1

    return status_count
```
