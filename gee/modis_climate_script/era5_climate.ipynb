```python
YEAR = 2024  # ← Có thể đổi thành 2020–2024
FOLDER_BASE_NAME = 'era5_climate_vietnam'
dates_to_process = dates_by_year[YEAR]

print(f"🔹 Năm {YEAR} có {len(dates_to_process)} ngày cần xử lý.")
```

```python
era5_land = ee.ImageCollection('ECMWF/ERA5_LAND/HOURLY')

for date_string in dates_to_process:
    print(f"\n🔸 Đang xử lý ngày {date_string}...")
    target_date = ee.Date(date_string)

    era5_for_day = era5_land.filterDate(target_date, target_date.advance(1, 'day'))
    image_count = era5_for_day.size().getInfo()
    print(f"   ➜ Số ảnh tìm thấy: {image_count}")

    if image_count == 0:
        print(f"⚠️ Không có dữ liệu cho ngày {date_string}. Bỏ qua.")
        continue

    # 1️⃣ Tính trung bình/ngày, trừ K → °C
    temp2m = era5_for_day.select('temperature_2m').map(lambda i: i.subtract(273.15)).mean().rename('temperature_2m')
    skin_temp = era5_for_day.select('skin_temperature').map(lambda i: i.subtract(273.15)).mean().rename('skin_temperature')
    soil_temp_l1 = era5_for_day.select('soil_temperature_level_1').map(lambda i: i.subtract(273.15)).mean().rename('soil_temperature_L1')
    soil_water_l1 = era5_for_day.select('volumetric_soil_water_layer_1').mean().rename('soil_water_L1')
    solar_rad = era5_for_day.select('surface_solar_radiation_downwards_hourly').sum().rename('surface_solar_radiation')
    precip = era5_for_day.select('total_precipitation_hourly').sum().multiply(1000).rename('total_precipitation')
    lai_low_veg = era5_for_day.select('leaf_area_index_low_vegetation').mean().rename('lai_low_veg')
    wind_u = era5_for_day.select('u_component_of_wind_10m').mean().rename('wind_u_10m')
    wind_v = era5_for_day.select('v_component_of_wind_10m').mean().rename('wind_v_10m')
    dewpoint = era5_for_day.select('dewpoint_temperature_2m').map(lambda i: i.subtract(273.15)).mean().rename('dewpoint_temp_2m')
    surface_pressure = era5_for_day.select('surface_pressure').mean().rename('surface_pressure')

    # 2️⃣ Kết hợp tất cả các band
    final_image = ee.Image.cat([
        temp2m, skin_temp, soil_temp_l1, soil_water_l1,
        solar_rad, precip, lai_low_veg, wind_u, wind_v,
        dewpoint, surface_pressure
    ]).clip(vietnamGeometry).toFloat()

    # 3️⃣ Xuất ảnh về Google Drive
    output_folder = f"{FOLDER_BASE_NAME}_{YEAR}"
    output_name = f"ERA5_Land_Daily_Vietnam_{date_string}"

    task = ee.batch.Export.image.toDrive(
        image=final_image,
        description=output_name,
        folder=output_folder,
        fileNamePrefix=output_name,
        region=vietnamGeometry.bounds(),
        scale=9000,
        crs='EPSG:4326',
        maxPixels=1e13
    )
    task.start()
    print(f"✅ Đã tạo tác vụ xuất: {output_name}")

print("\n🎯 Hoàn tất việc tạo tác vụ. Kiểm tra tab 'Tasks' trong Earth Engine Dashboard.")
print("🔗 Theo dõi: https://code.earthengine.google.com/tasks")
```
