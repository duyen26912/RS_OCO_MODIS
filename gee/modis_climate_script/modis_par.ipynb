```python
YEAR = 2024 # ← Có thể đổi thành 2020–2024
FOLDER_BASE_NAME = 'modis_par_vietnam'

dates_to_process = dates_by_year[YEAR]
print(f"🔹 Năm {YEAR} có {len(dates_to_process)} ngày cần xử lý.")
```

```python
mcd18c2 = ee.ImageCollection('MODIS/062/MCD18C2')

def process_par_band(band):
    """Giữ lại các giá trị PAR hợp lệ trong dải 0–1500"""
    return band.updateMask(band.gte(0).And(band.lte(1500)))

for date_string in dates_to_process:
    print(f"\n🔸 Đang xử lý ngày {date_string}...")

    start_date = f"{date_string}T00:00:00"
    end_date = f"{date_string}T23:59:59"

    filtered = mcd18c2.filterDate(start_date, end_date).filterBounds(vietnamGeometry)
    count = filtered.size().getInfo()
    print(f"   ➜ Số ảnh MCD18C2 tìm thấy: {count}")

    if count == 0:
        print(f"⚠️ Không tìm thấy dữ liệu cho ngày {date_string}. Bỏ qua.")
        continue

    # 1️⃣ Tạo mosaic và cắt biên giới
    image = filtered.mosaic().clip(vietnamGeometry)

    # 2️⃣ Chọn các band PAR
    selected = image.select(par_band_names)

    # 3️⃣ Xử lý từng band
    processed_bands = [process_par_band(selected.select(name)) for name in par_band_names]

    # 4️⃣ Kết hợp các band đã xử lý
    final_image = ee.Image.cat(processed_bands).toFloat()

    # 5️⃣ Xuất ảnh về Google Drive
    output_folder = f"{FOLDER_BASE_NAME}_{YEAR}"
    output_name = f"MCD18C2_Vietnam_PAR_4Times_{date_string}"

    task = ee.batch.Export.image.toDrive(
        image=final_image,
        description=output_name,
        folder=output_folder,
        fileNamePrefix=output_name,
        region=vietnamGeometry.bounds(),
        scale=5600,
        crs='EPSG:4326',
        maxPixels=1e13
    )
    task.start()
    print(f"✅ Đã tạo tác vụ xuất: {output_name}")

print("\n🎯 Hoàn tất việc tạo tác vụ. Kiểm tra tab 'Tasks' trong Earth Engine Dashboard.")
print("🔗 Theo dõi: https://code.earthengine.google.com/tasks")
```
