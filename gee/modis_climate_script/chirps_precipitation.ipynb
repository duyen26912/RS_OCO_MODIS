```python
# Lấy biên giới Việt Nam
countries = ee.FeatureCollection("FAO/GAUL/2015/level0")
vietnam = countries.filter(ee.Filter.eq('ADM0_NAME', 'Viet Nam')).first()
vietnam_geometry = vietnam.geometry()
```

```python
# Tham số hiển thị (tùy chọn cho trực quan, không bắt buộc)
precipitation_vis_params = {
    'min': 0.0,
    'max': 50.0,
    'palette': [
        '#FFFFFF', '#EFF3FF', '#BDDEBB', '#78C679', '#41AB5D', '#238443', '#005A32',
        '#FFFF00', '#FE9929', '#EC7014', '#CC4C02', '#993404', '#662506'
    ]
}
```

```python
# CHỌN NĂM CẦN XỬ LÝ
YEAR = 2024  # ← Có thể đổi: 2020, 2021, 2022, 2023, 2024

# CHỌN TÊN FOLDER TRÊN GOOGLE DRIVE (tuỳ chọn)
FOLDER_BASE_NAME = 'chirps_precipitation'
dates_to_process = dates_by_year[YEAR]

print(f"🔹 Năm {YEAR} có {len(dates_to_process)} ngày cần xử lý.")
```

```python
# Xử lý từng ngày
for date_str in dates_to_process:
    print(f"🔹 Đang xử lý ngày: {date_str}")
    start_date = f"{date_str}T00:00:00"
    end_date = f"{date_str}T23:59:59"

    chirps = (ee.ImageCollection('UCSB-CHG/CHIRPS/DAILY')
              .filterDate(start_date, end_date)
              .filterBounds(vietnam_geometry))

    image_count = chirps.size().getInfo()
    if image_count == 0:
        print(f"⚠️ Không có dữ liệu cho {date_str}, bỏ qua.")
        continue

    # Nếu có dữ liệu → lấy ảnh và xử lý
    chirps_image = chirps.mosaic().clip(vietnam_geometry)
    precipitation = chirps_image.select('precipitation')
    processed_precip = precipitation.updateMask(precipitation.gte(0))
    final_image = processed_precip.toFloat()

    # Xuất sang Google Drive
    task = ee.batch.Export.image.toDrive(
        image=final_image,
        description=f'CHIRPS_Vietnam_Precipitation_{date_str}',
        folder=f'{FOLDER_BASE_NAME}_{YEAR}',
        fileNamePrefix=f'CHIRPS_Vietnam_Precipitation_{date_str}',
        region=vietnam_geometry.bounds(),
        scale=5566,
        crs='EPSG:4326',
        maxPixels=1e13
    )
    task.start()
    print(f"✅ Đã tạo tác vụ xuất dữ liệu cho ngày {date_str}.")
print("🔗 Theo dõi: https://code.earthengine.google.com/tasks")
```
