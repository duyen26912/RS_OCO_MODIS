## Xử lý dữ liệu raster bản đồ lớp phủ đất (Land Cover Map) sang bản đồ nhị phân chỉ chứa thông tin về diện tích trồng lúa.

### Import thư viện

```python
import rasterio # Dùng để đọc dữ liệu raster (.tif)
from rasterio.windows import Window # Dùng để xử lý cửa sổ raster
import numpy as np # Dùng để xử lý mảng số học
```

### Chuyển đổi bản đồ lớp phủ đất sang bản đồ nhị phân chỉ chứa giá trị lúa gạo

```python
# Chuyển đổi giá trị lớp phủ đất lúa gạo sang bản đồ nhị phân
RICE_VALUE = 3
input_raster = r"E:\RanhGioi\LandCoverMap\lc_vn_10m_wgs84.tif"
output_raster = r"E:\RanhGioi\LandCoverMap\lc_vn_10m_wgs84_rice_binary.tif"

with rasterio.open(input_raster) as src:
    profile = src.profile
    profile.update(dtype=rasterio.uint8)

    with rasterio.open(output_raster, 'w', **profile) as dst:
        for ji, window in src.block_windows(1):
            data = src.read(1, window=window)
            new_data = np.where(data == RICE_VALUE, 1, 0).astype(np.uint8)
            dst.write(new_data, 1, window=window)
            print(f"Processed windo {ji}")
```
