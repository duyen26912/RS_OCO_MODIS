## Kiểm tra dữ liệu CO₂ bản đồ lớp phủ đất (Land Cover) và ranh giới Viêt Nam để xác định hệ tọa độ không gian

### Import thư viện


```python
import geopandas as gpd # Dùng để xử lý dữ liệu không gian (shapefile, geodataframe)
from shapely.geometry import Point # Dùng để tạo đối tượng điểm từ tọa độ
import pandas as pd # Dùng để xử lý dữ liệu bảng (dataframe)
import rasterio # Dùng để đọc dữ liệu raster (.tif)
```

### Đọc thông tin của các dữ liệu

```python
# --- Đọc thông tin raster (.tif) ---
raster_path = r"E:\RanhGioi\LandCoverMap\landcover_VNM_250m_utm.tif"
with rasterio.open(raster_path) as src:
    print("File:", raster_path)
    print("CRS (hệ tọa độ):", src.crs)
    print("Độ phân giải (resolution):", src.res)
    print("Kích thước (width x height):", src.width, "x", src.height)
    print()


# --- Đọc thông tin shapefile (.shp) ---
shapefile_path = r"E:\RanhGioi\VNM_adm\gadm41_VNM_0.shp"
shp = gpd.read_file(shapefile_path)
print("File:", shapefile_path)
print("CRS (hệ tọa độ):", shp.crs)
print("Số lượng đối tượng (features):", len(shp))
```

### Đọc dữ liệu CO₂ từ file CSV sau đó chuyển thành GeoDataFrame để xử lý không gian.

```python

df = pd.read_csv(r"E:\DownloadData\co2_nasa\data_processed\csv_yearly\2020\oco2\oco2_vn_qual0_2020.csv")

# Tạo GeoDataFrame từ các cột tọa độ
gdf = gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(df.longitude, df.latitude))

# Nếu bạn biết hoặc muốn kiểm tra, tạm gán CRS là WGS84
gdf.set_crs("EPSG:4326", inplace=True)

# Kiểm tra CRS
print("Hệ tọa độ hiện tại:", gdf.crs)
```
